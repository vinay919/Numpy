```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],    # Paradise Biryanig
    [2, 120000, 140000, 160000, 190000],    # Beijing Bites
    [3, 200000, 230000, 260000, 300000],    # Pizza Hub
    [4, 180000, 210000, 240000, 270000],    # Burger Point
    [5, 160000, 185000, 205000, 230000]     # Chai Point
])

print("====== Zomata sales analysis =======")
print("\n Sales data shape:", sales_data.shape)
print("\n Sample data for 1st three restaurant:",sales_data[:3])
```

## total sales per year

```python
print("Total data year wise:",np.sum(sales_data, axis=0))
```

```python
print("Total data restaurant wise:",np.sum(sales_data[:,1:], axis=1))
```

```python
min_sales = np.min(sales_data[:,1:],axis=1)
print("Minimum sales:",min_sales)
```

```python
max_sales = np.max(sales_data[:,1:], axis=1)
print("Maximum Sales:",max_sales)
```

```python
avg_sales = np.average(sales_data[:,1:], axis=1)
print("Average sales:",avg_sales)
```

```python
# cumulative 
cumsum = np.cumsum(sales_data[:,1:], axis=1)
print("Cumulative Sales:",cumsum)
```

## Matplot lib


```python
plt.figure(figsize=(9,6))
plt.plot(np.mean(cumsum, axis=0))
plt.title("Average cumulative sales accross all restaurant")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
```

## Vector

```python
vector1 = np.array([1,2,3,4,5])
vector2 = np.array([6,7,8,9,10])

print("Vector addition",vector1+vector2)
print("Vector multiplication",vector1*vector2)
print("Vector Dot product",np.dot(vector1,vector2))

angle = np.arccos(np.dot(vector1,vector2)/(np.linalg.norm(vector1)*np.linalg.norm(vector2)))
print("Angle",angle)
```

## Vectorization

```python
restaurant_types = np.array(['biryani','chineese','pizza','burger','cafe'])
vectorized_upper = np.vectorize(str.upper)
print("Without Vectorized Upper",restaurant_types)
print("Vectorized Upper",vectorized_upper(restaurant_types))
```
